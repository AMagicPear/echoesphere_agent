```python
from phoenix.otel import register
from openinference.instrumentation.smolagents import SmolagentsInstrumentor

register(project_name="echoesphere-debug", set_global_tracer_provider=True)
SmolagentsInstrumentor().instrument()
```

```python
import logging
from echoesphere_agent.agent import DecisionAgent
import dotenv
import os

dotenv.load_dotenv("../.env")
logger = logging.getLogger("echoesphere.agent")
logger.setLevel(logging.INFO)

decision_agent = DecisionAgent(
    model_id="minimax/MiniMax-M2.7",
    api_key=os.getenv("MINIMAX_API_KEY"),
    api_base=os.getenv("MINIMAX_API_BASE"),
)
with open("./system_prompt.txt", "w") as f:
    f.write(decision_agent.agent.memory.system_prompt.system_prompt)
```

```python
decision_agent.agent.run("请设置环境效果为雾气")
```
